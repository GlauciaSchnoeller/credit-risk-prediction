```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

from src.evaluation import (
    compare_models,
    get_confusion_matrix,
)
from src.models import (
    create_decision_tree,
    create_logistic_regression,
    create_random_forest,
)

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "credit_default.csv"

df = pd.read_csv(DATA_PATH)

X = df.drop(columns=["Y"])
y = df["Y"]

print("Dataset shape:", df.shape)
print("Features:", X.shape)
print("Target:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
models = {
    "Logistic Regression": create_logistic_regression(),
    "Decision Tree": create_decision_tree(),
    "Random Forest": create_random_forest()
}
```

```python
for model_name, model in models.items():

    model.fit(X_train, y_train)

    print(f"{model_name} trained successfully!")
```

```python
results_df = compare_models(
    models,
    X_test,
    y_test
)

display(results_df.round(4))
```

```python
logistic_cm = get_confusion_matrix(
    models["Logistic Regression"],
    X_test,
    y_test
)

print(logistic_cm)
```

# Initial Model Evaluation
- Random Forest achieved the best overall performance
- Random Forest obtained the highest Accuracy (0.8148), F1-score (0.4635), and ROC-AUC (0.7546)
- Logistic Regression achieved the highest Precision (0.6923)
- Decision Tree achieved the highest Recall (0.3986)
- All models presented relatively low Recall values
